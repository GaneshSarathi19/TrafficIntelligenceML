### 2.3 Detect possible delay outliers with the IQR rule

The interquartile range (IQR) is `Q3 - Q1`. Values below `Q1 - 1.5 × IQR` or above `Q3 + 1.5 × IQR` are flagged for inspection. They are reported here rather than automatically deleted.


```python
Q1 = df["delay"].quantile(0.25)
Q3 = df["delay"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = df[
    (df["delay"] < lower_limit) |
    (df["delay"] > upper_limit)
]

print("Number of outliers:", len(outliers))
display(outliers)
```

## 4. Preprocessing the clustering features

The clustering algorithms need numeric inputs. We select the traffic features, encode categories, and scale the resulting matrix. The `congestion` label is deliberately excluded from `X`.


```python
# ==============================
# PREPROCESSING
# ==============================

# Features used for clustering
features = ["road_id", "hour", "day", "delay"]

X = df[features].copy()

print("Features selected for clustering:")
print(X.columns.tolist())

display(X.head())
```

### 4.1 Inspect selected feature data types

Checking the types before encoding helps identify which features need conversion into numeric columns.


```python
print("Data types:")
print(X.dtypes)
```

### 4.2 One-hot encode categorical features

`pd.get_dummies()` converts `road_id` and `day` into indicator columns containing 0/1 values. This makes these categories usable by the clustering algorithms.


```python
X_encoded = pd.get_dummies(
    X,
    columns=["road_id", "day"],
    dtype=int
)

print("Shape after encoding:", X_encoded.shape)

display(X_encoded.head())
```

### 4.3 Review the encoded feature matrix

This cell lists the generated columns and checks their data types after one-hot encoding.


```python
print("Encoded columns:")
print(X_encoded.columns.tolist())

print("\nData types:")
print(X_encoded.dtypes)
```

`StandardScaler()` standardizes each feature using its mean and standard deviation. Scaling is important because distance-based clustering can otherwise be dominated by features with larger numeric ranges.

### 4.5 Preview standardized values

## 5. Principal Component Analysis (PCA)

PCA transforms the standardized features into principal components ordered by how much variance they explain. In this notebook, PCA is used to study information retention and to create 2D visualizations.


### 5.1 Import PCA tools

PCA is a dimensionality-reduction method. It creates new components that summarize variation in the scaled features.


```python
from sklearn.decomposition import PCA
import numpy as np
import matplotlib.pyplot as plt
```

### 5.2 Fit PCA and calculate explained variance

`PCA()` is fitted to the scaled matrix. The explained variance ratio reports the proportion of variation captured by each principal component; the cumulative sum shows the total captured by the first several components.


```python
pca = PCA()

X_pca = pca.fit_transform(X_scaled)

explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

print("Original number of features:", X_scaled.shape[1])
print("Number of principal components:", X_pca.shape[1])

print("\nExplained variance ratio:")
print(explained_variance.round(4))

print("\nCumulative explained variance:")
print(cumulative_variance.round(4))
```

### 5.3 Plot cumulative explained variance

The curve helps show how many principal components are needed to retain a chosen proportion of the data's variance. The dashed lines mark 90% and 95% targets.


```python
plt.figure(figsize=(10, 5))

plt.plot(
    range(1, len(explained_variance) + 1),
    cumulative_variance,
    marker="o"
)

plt.axhline(y=0.90, linestyle="--", label="90% variance")
plt.axhline(y=0.95, linestyle="--", label="95% variance")

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA - Cumulative Explained Variance")
plt.xticks(range(1, len(explained_variance) + 1))
plt.grid(True)
plt.legend()
plt.show()
```

### 5.4 Count components needed for 90% and 95% variance

`np.argmax()` locates the first component count at which the cumulative variance reaches each threshold.


```python
components_90 = np.argmax(cumulative_variance >= 0.90) + 1
components_95 = np.argmax(cumulative_variance >= 0.95) + 1

print("Components needed for 90% variance:", components_90)
print("Components needed for 95% variance:", components_95)
```

### 5.5 Create a two-component PCA projection

This 2D representation is used for cluster plots. It is a visualization aid; it does not necessarily preserve all the information used by the clustering models.


```python
pca_2d = PCA(n_components=2)

X_pca_2d = pca_2d.fit_transform(X_scaled)

print("2D PCA shape:", X_pca_2d.shape)
print(
    "Variance explained by 2 components:",
    round(pca_2d.explained_variance_ratio_.sum() * 100, 2),
    "%"
)
```

## 6. K-Means Clustering

K-Means partitions observations into a chosen number of clusters by minimizing the within-cluster sum of squared distances. The Elbow Method is used first to select a reasonable candidate value for `k`; the resulting clusters are then evaluated and interpreted.


### 6.1 Choose a candidate cluster count with the Elbow Method

K-Means is fitted for values of `k` from 1 to 10. Inertia measures the within-cluster sum of squared distances. Look for the point where adding more clusters starts producing smaller improvements.


```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Test different numbers of clusters
k_values = range(1, 11)
inertias = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    inertias.append(kmeans.inertia_)

# Plot the Elbow curve
plt.figure(figsize=(8, 5))
plt.plot(k_values, inertias, marker="o")

plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("Elbow Method for K-Means Clustering")
plt.xticks(list(k_values))
plt.grid(True)
plt.show()
```

### 6.2 Fit K-Means and assign cluster labels

The model uses `k=7` based on the current Elbow plot. `random_state=42` makes the initialization reproducible, and `n_init=10` tries multiple initial centroid sets. `fit_predict()` fits the model and returns a cluster ID for each row.


```python
from sklearn.cluster import KMeans
import pandas as pd

# Train the K-Means model
kmeans = KMeans(
    n_clusters=7,
    random_state=42,
    n_init=10
)

# Assign a cluster to each observation
kmeans_labels = kmeans.fit_predict(X_scaled)

# Store cluster assignments for interpretation
df_kmeans = df.copy()
df_kmeans["kmeans_cluster"] = kmeans_labels

# Display the number of observations in each cluster
print("Cluster Counts:")
print(df_kmeans["kmeans_cluster"].value_counts().sort_index())

print("\nDataset with Cluster Labels:")
display(df_kmeans.head())
```

### 6.3 Evaluate K-Means

The Silhouette Score is generally better when higher; the Davies–Bouldin Index is generally better when lower; and the Calinski–Harabasz Index is generally better when higher. Interpret them together rather than relying on one score alone.


```python
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

# Calculate clustering evaluation metrics
silhouette = silhouette_score(X_scaled, kmeans_labels)
davies_bouldin = davies_bouldin_score(X_scaled, kmeans_labels)
calinski_harabasz = calinski_harabasz_score(X_scaled, kmeans_labels)

print("K-Means Evaluation (k=7)")
print("-" * 35)
print(f"Silhouette Score:          {silhouette:.4f}")
print(f"Davies-Bouldin Index:      {davies_bouldin:.4f}")
print(f"Calinski-Harabasz Index:   {calinski_harabasz:.4f}")
```

### 6.4 Visualize K-Means labels in 2D PCA space

Each point is one traffic observation, colored by its K-Means cluster. The clustering used all scaled features; PCA is only used here to display the groups in two dimensions.


```python
import matplotlib.pyplot as plt

# Visualize K-Means clusters using the existing 2D PCA projection
plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    X_pca_2d[:, 0],
    X_pca_2d[:, 1],
    c=kmeans_labels,
    cmap="tab10",
    s=40,
    alpha=0.8
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clustering Visualization using PCA (k=7)")

colorbar = plt.colorbar(scatter)
colorbar.set_label("Cluster")

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 6.5 Describe the K-Means clusters

The tables summarize each cluster's size, delay, and hour, then show the congestion categories within each group. Congestion is used only after clustering for interpretation.


```python
# Summarize traffic characteristics for each K-Means cluster
cluster_summary = df_kmeans.groupby("kmeans_cluster").agg(
    observations=("delay", "size"),
    average_delay=("delay", "mean"),
    median_delay=("delay", "median"),
    average_hour=("hour", "mean")
).round(3)

print("K-Means Cluster Summary:")
display(cluster_summary)

# Count congestion categories in each cluster
congestion_counts = pd.crosstab(
    df_kmeans["kmeans_cluster"],
    df_kmeans["congestion"]
)

print("Congestion Counts by Cluster:")
display(congestion_counts)

# Percentage distribution of congestion within each cluster
congestion_percentages = (
    pd.crosstab(
        df_kmeans["kmeans_cluster"],
        df_kmeans["congestion"],
        normalize="index"
    ) * 100
).round(2)

print("Congestion Distribution (%) by Cluster:")
display(congestion_percentages)
```

### 6.6 Plot congestion percentages by K-Means cluster

The stacked bars compare the percentage composition of congestion categories within each cluster, making cluster profiles easier to explain.


```python
# Plot congestion distribution within each K-Means cluster
congestion_order = ["FREE", "LIGHT", "MODERATE", "HEAVY"]

plot_data = congestion_percentages.reindex(
    columns=congestion_order,
    fill_value=0
)

ax = plot_data.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 6)
)

ax.set_xlabel("K-Means Cluster")
ax.set_ylabel("Observations (%)")
ax.set_title("Congestion Distribution by K-Means Cluster")
ax.legend(title="Congestion Level", bbox_to_anchor=(1.02, 1),
          loc="upper left")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 7. Agglomerative Hierarchical Clustering

Agglomerative clustering builds a hierarchy by repeatedly merging the closest groups. This section creates a dendrogram, compares linkage strategies, evaluates the resulting clusters, and interprets Ward clusters using the original traffic columns.


### 7.1 Build a Ward-linkage dendrogram

Agglomerative clustering starts with observations as separate groups and merges groups step by step. Ward linkage chooses merges that minimize the increase in within-cluster variation. The truncated dendrogram keeps the plot readable.


```python
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram

# Use the standardized features from preprocessing
# Congestion is excluded from model training
Z = linkage(X_scaled, method="ward", metric="euclidean")

# Plot a readable, truncated dendrogram
plt.figure(figsize=(14, 6))

dendrogram(
    Z,
    truncate_mode="lastp",
    p=30,
    show_leaf_counts=True
)

plt.title("Agglomerative Hierarchical Clustering Dendrogram (Ward Linkage)")
plt.xlabel("Cluster Groups")
plt.ylabel("Ward Linkage Distance")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

### 7.2 Compare four linkage strategies

Ward, complete, average, and single linkage define the distance between groups differently. The four dendrograms provide a visual comparison of their merge structures.


```python
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt

linkage_methods = ["ward", "complete", "average", "single"]

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

for method, ax in zip(linkage_methods, axes.flatten()):
    Z_method = linkage(
        X_scaled,
        method=method,
        metric="euclidean"
    )

    dendrogram(
        Z_method,
        truncate_mode="lastp",
        p=30,
        show_leaf_counts=True,
        ax=ax
    )

    ax.set_title(f"{method.capitalize()} Linkage")
    ax.set_xlabel("Cluster Groups")
    ax.set_ylabel("Merge Distance")
    ax.grid(axis="y", alpha=0.25)

plt.suptitle(
    "Agglomerative Clustering: Linkage Strategy Comparison",
    fontsize=14
)
plt.tight_layout()
plt.show()
```

### 7.3 Evaluate linkage strategies at seven clusters

For an initial apples-to-apples comparison, each linkage strategy is evaluated with `n_clusters=7`. The same three clustering metrics are calculated for each strategy.


```python
import pandas as pd
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

linkage_methods = ["ward", "complete", "average", "single"]
agglo_results = []
agglo_labels = {}

for method in linkage_methods:
    model = AgglomerativeClustering(
        n_clusters=7,
        linkage=method,
        metric="euclidean"
    )

    labels = model.fit_predict(X_scaled)
    agglo_labels[method] = labels

    agglo_results.append({
        "Linkage": method.capitalize(),
        "Clusters": len(set(labels)),
        "Silhouette Score": silhouette_score(X_scaled, labels),
        "Davies-Bouldin Index": davies_bouldin_score(X_scaled, labels),
        "Calinski-Harabasz Index": calinski_harabasz_score(
            X_scaled, labels
        )
    })

agglo_results_df = pd.DataFrame(agglo_results).round(4)

print("Agglomerative Clustering: Linkage Comparison (k=7)")
display(agglo_results_df)
```

### 7.4 Verify cluster assignments with Adjusted Rand Index (ARI)

ARI compares two sets of cluster assignments while ignoring the numeric names assigned to clusters. A value of 1.0 indicates identical partitions; a value near 0 indicates agreement similar to random assignment.


```python
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score,
    adjusted_rand_score
)
import pandas as pd

linkage_methods = ["ward", "complete", "average", "single"]
agglo_labels_check = {}

# Refit each model and check its cluster assignments
for method in linkage_methods:
    model = AgglomerativeClustering(
        n_clusters=7,
        linkage=method,
        metric="euclidean"
    )

    labels = model.fit_predict(X_scaled)
    agglo_labels_check[method] = labels

    print(f"\n{method.upper()} LINKAGE")
    print("Cluster sizes:")
    print(pd.Series(labels).value_counts().sort_index().to_dict())

    print("Silhouette:", f"{silhouette_score(X_scaled, labels):.8f}")
    print("Davies-Bouldin:", f"{davies_bouldin_score(X_scaled, labels):.8f}")
    print("Calinski-Harabasz:", f"{calinski_harabasz_score(X_scaled, labels):.8f}")

# Compare whether the algorithms produce the same groupings
print("\nPAIRWISE CLUSTER AGREEMENT (Adjusted Rand Index)")
for i, method_a in enumerate(linkage_methods):
    for method_b in linkage_methods[i + 1:]:
        score = adjusted_rand_score(
            agglo_labels_check[method_a],
            agglo_labels_check[method_b]
        )
        print(f"{method_a} vs {method_b}: {score:.4f}")

print(
    "\nWard vs K-Means:",
    f"{adjusted_rand_score(kmeans_labels, agglo_labels_check['ward']):.4f}"
)
```

### 7.5 Visualize Ward clusters in 2D PCA space

This plot colors the same 2D PCA projection using the Ward cluster assignments. As with K-Means, it is a two-dimensional view of clustering performed on the full scaled feature set.


```python
# Use the Ward linkage cluster assignments
ward_labels = agglo_labels_check["ward"]

# Store cluster assignments for interpretation
df_agglo = df.copy()
df_agglo["agglo_cluster"] = ward_labels

# Visualize clusters using the existing 2D PCA projection
plt.figure(figsize=(9, 6))

scatter = plt.scatter(
    X_pca_2d[:, 0],
    X_pca_2d[:, 1],
    c=ward_labels,
    cmap="tab10",
    s=40,
    alpha=0.8
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Agglomerative Clustering using PCA (Ward Linkage, k=7)")

colorbar = plt.colorbar(scatter)
colorbar.set_label("Cluster")

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### 7.6 Interpret Ward clusters with the original traffic data

The summary describes delay and hour in each Ward cluster. Congestion counts and percentages are used after fitting to explain the traffic patterns in the groups.


```python
# Summarize clusters using the original traffic data
ward_summary = df_agglo.groupby("agglo_cluster").agg(
    observations=("delay", "size"),
    average_delay=("delay", "mean"),
    median_delay=("delay", "median"),
    average_hour=("hour", "mean")
).round(3)

print("Ward Cluster Summary:")
display(ward_summary)

# Use congestion only for post-clustering interpretation
ward_congestion = pd.crosstab(
    df_agglo["agglo_cluster"],
    df_agglo["congestion"]
)

print("Congestion Counts by Ward Cluster:")
display(ward_congestion)

ward_congestion_pct = (
    pd.crosstab(
        df_agglo["agglo_cluster"],
        df_agglo["congestion"],
        normalize="index"
    ) * 100
).round(2)

print("Congestion Distribution (%) by Ward Cluster:")
display(ward_congestion_pct)
```

## 8. Compare K-Means and Agglomerative Clustering

This final metrics table compares the two required algorithms using the same scaled features and number of clusters. The ARI below the table checks whether their actual group assignments agree.


```python
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score,
    adjusted_rand_score
)
import pandas as pd

# Compare the two required clustering algorithms
models = {
    "K-Means": kmeans_labels,
    "Agglomerative (Ward)": ward_labels
}

comparison_rows = []

for model_name, labels in models.items():
    comparison_rows.append({
        "Algorithm": model_name,
        "Clusters": len(set(labels)),
        "Silhouette Score": silhouette_score(X_scaled, labels),
        "Davies-Bouldin Index": davies_bouldin_score(
            X_scaled, labels
        ),
        "Calinski-Harabasz Index": calinski_harabasz_score(
            X_scaled, labels
        )
    })

comparison_df = pd.DataFrame(comparison_rows)

print("Final Clustering Algorithm Comparison")
display(comparison_df.round(4))

print(
    "\nAdjusted Rand Index (K-Means vs Ward):",
    round(adjusted_rand_score(kmeans_labels, ward_labels), 4)
)
```
